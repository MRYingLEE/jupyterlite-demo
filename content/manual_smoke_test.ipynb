# A2UI JupyterLab manual smoke test

Run the form cell. While it waits, enter a name and click **Submit**. The cell will return the action and update the status. Later cells wait in the kernel's execution queue until the form cell finishes.

If a saved output reports that its widget model was not found, clear the output and rerun the form cell. See the [README](../README.md#saved-widget-output) for saving widget state.

```python
from IPython.display import display

from a2ui_jupyter import A2UISurface

surface = A2UISurface()
surface.send(
    [
        {
            "version": "1.0",
            "createSurface": {
                "surfaceId": "manual-smoke-test",
                "catalogId": "https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json",
            },
        },
        {
            "version": "1.0",
            "updateDataModel": {
                "surfaceId": "manual-smoke-test",
                "path": "/person/name",
                "value": "",
            },
        },
        {
            "version": "1.0",
            "updateDataModel": {
                "surfaceId": "manual-smoke-test",
                "path": "/status",
                "value": "Waiting for submission",
            },
        },
        {
            "version": "1.0",
            "updateComponents": {
                "surfaceId": "manual-smoke-test",
                "components": [
                    {
                        "id": "root",
                        "component": "Column",
                        "children": ["title", "name", "submit", "status"],
                        "align": "stretch",
                    },
                    {
                        "id": "title",
                        "component": "Text",
                        "text": "A2UI JupyterLab smoke test",
                    },
                    {
                        "id": "name",
                        "component": "TextField",
                        "label": "Name",
                        "placeholder": "Enter a name",
                        "value": {"@path": "/person/name"},
                    },
                    {
                        "id": "submit",
                        "component": "Button",
                        "child": "submit_label",
                        "variant": "primary",
                        "action": {
                            "event": {
                                "name": "submit",
                                "context": {
                                    "name": {"@path": "/person/name"},
                                },
                            },
                        },
                    },
                    {
                        "id": "submit_label",
                        "component": "Text",
                        "text": "Submit",
                    },
                    {
                        "id": "status",
                        "component": "Text",
                        "text": {"@path": "/status"},
                    },
                ],
            },
        },
    ]
)

display(surface)

action = await surface.next_action()
action_data = action.model_dump(by_alias=True, exclude_unset=True)
surface.send(
    {
        "version": "1.0",
        "updateDataModel": {
            "surfaceId": "manual-smoke-test",
            "path": "/status",
            "value": f"Submitted: {action_data['action']['context']['name']}",
        },
    }
)
action_data
```

After submitting, run this cell to inspect the action again.

```python
action_data
```

The form cell already updated the status. Run this cell to send the same update again.

```python
surface.send(
    {
        "version": "1.0",
        "updateDataModel": {
            "surfaceId": "manual-smoke-test",
            "path": "/status",
            "value": f"Submitted: {action_data['action']['context']['name']}",
        },
    }
)
```
